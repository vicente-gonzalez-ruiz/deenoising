```python
X_R = skimage_io.imread(args.R)
X_G = skimage_io.imread(args.G)
X_B = skimage_io.imread(args.B)
```

```python
plt.title(r"$\mathrm{TwoPhoton\_BPAE\_R}$")
plt.imshow(utils.normalize(X_R), cmap="gray")
plt.savefig(args.output1, bbox_inches='tight')
```

```python
plt.title(r"$\mathrm{TwoPhoton\_BPAE\_G}$")
plt.imshow(utils.normalize(X_G), cmap="gray")
plt.savefig(args.output2, bbox_inches='tight')
```

```python
plt.title(r"$\mathrm{TwoPhoton\_BPAE\_B}$")
plt.imshow(utils.normalize(X_B), cmap="gray")
plt.savefig(args.output3, bbox_inches='tight')
```

```python
X = np.stack((X_R, X_G, X_B), axis=-1)
```

```python
plt.title(r"$\mathrm{TwoPhoton\_BPAE}$")
X = utils.RGB_normalize(X)
plt.imshow(X)
plt.savefig(args.output4, bbox_inches='tight')
```
