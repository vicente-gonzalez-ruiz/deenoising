## Poisson noise (shot noise)
\begin{equation}
  Y=\frac{N}{\gamma},~N∼\mathcal{P}(λ=\gamma X),
\end{equation}
where:
* $X\ge 0$.
* $\lambda\in{\mathbb R}\ge 0$ is the rate of the Poisson distribution, resulting that $\lambda={\mathbb E}(N)=\text{Var}(N)$.
