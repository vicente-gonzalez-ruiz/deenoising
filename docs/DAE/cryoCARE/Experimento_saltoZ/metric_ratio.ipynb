```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("metrics_ratio.txt", delim_whitespace=True)

print(df.columns.tolist())

plt.figure(figsize=(4*1.5, 3*1.5))

colors = ["darkorange", "purple", "teal"]

for column, color in zip(df.columns[1:], colors):
    plt.plot(
        df["#thickness"],
        df[column],
        marker="o",
        linewidth=3,
        color=color,
        label=column.replace("Ratio_", "")
    )

plt.xlabel("Slice Thickness")
plt.ylabel("N2N-REO/N2N-EO Ratio")
plt.legend()
plt.grid(True, linestyle="--", alpha=0.5)
plt.tight_layout()
plt.show()
```
