```python
import mrcfile
from collections import namedtuple
import numpy as np
```

```python
Args = namedtuple("args", ["input", "even_output", "odd_output"])
args = Args("/home/jupyter-jjfdez/tomograms/CNB_20240305_stack_ali_crop.mrc", "even.mrc", "odd.mrc")
#args = Args("/home/jupyter-vruiz/gdrive_TomogramDenoising/tomograms/empiar_11275_small.mrc", "even_small.mrc", "odd_small.mrc")
```

```python
vol_MRC = mrcfile.open(args.input)
noisy = vol_MRC.data
print(f"noisy.shape={noisy.shape}")

#slice_shape = (1, noisy.shape[1], noisy.shape[2])
#new_slice = np.full(slice_shape, 128, dtype=noisy.dtype)

#even = np.concatenate((new_slice, noisy[0::2,:,:], new_slice, new_slice))
#odd = np.concatenate((new_slice, new_slice, noisy[1::2,:,:], new_slice, new_slice))

even = noisy[0::2,:,:]
odd = noisy[1::2,:,:]

print("Writing", args.even_output)
with mrcfile.new(args.even_output, overwrite=True) as mrc:
    print(f"even.shape={even.shape}")
    mrc.set_data(even)
    mrc.data

print("Writing", args.odd_output)
with mrcfile.new(args.odd_output, overwrite=True) as mrc:
    print(f"odd.shape={odd.shape}")
    mrc.set_data(odd)
    mrc.data
```
