```python
def SSIM(original_fn, denoised_fn):
    print(f"Original: {original_fn}")
    print(f"Denoised: {denoised_fn}")
    with mrcfile.open(original_fn, permissive=True) as original, mrcfile.open(denoised_fn, permissive=True) as denoised:
        # Load data as float32 to prevent overflow issues during subtraction
        original_data = original.data.astype(np.float32)
        denoised_data = denoised.data.astype(np.float32)
    SSIM = skimage.metrics.structural_similarity(original_data, denoised_data, data_range=original_data.max() - original_data.min()) #IT.distortion.SSIM(original_data.astype(np.uint8), denoised_data)
    print(f"SSIM={SSIM:.2f}")
```

```python
import numpy as np
import mrcfile

def rawread_MRC(fn):
    with mrcfile.open(fn, permissive=True) as m:
        nx = m.header.nx
        ny = m.header.ny
        nz = m.header.nz
        nsymbt = m.header.nsymbt

    # compute data offset
    offset = 1024 + nsymbt

    # read raw bytes manually
    data = np.fromfile(
        fn,
        dtype=np.float32,   # <-- force desired dtype
        offset=offset
    )

    # reshape manually
    return data.reshape((nz, ny, nx))
```

```python
def read_MRC(fn):
    with mrcfile.open(fn, permissive=True) as data:
        return data.data
```

```python
def array_SSIM(original, denoised):
    SSIM = skimage.metrics.structural_similarity(original, denoised, data_range=original.max() - original.min())
    print(f"SSIM={SSIM:.2f}")
```
