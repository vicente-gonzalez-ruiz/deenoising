```python
import os
import glob
import numpy as np
import mrcfile
import pandas as pd
from collections import namedtuple
import scipy.stats
```

```python
def PCC(original_fn, denoised_fn):
    print(f"Original: {original_fn}")
    print(f"Denoised: {denoised_fn}")
    with mrcfile.open(original_fn, permissive=True) as original, mrcfile.open(denoised_fn, permissive=True) as denoised:
        # Load data as float32 to prevent overflow issues during subtraction
        original_data = original.data.astype(np.float32)
        denoised_data = denoised.data.astype(np.float32)
    PCC = scipy.stats.pearsonr(original_data.flatten(), denoised_data.flatten())[0]
    print(f"PCC={PCC:.2f}")
```

```python
import numpy as np
import mrcfile

def rawread_MRC(fn):
    with mrcfile.open(fn, permissive=True) as m:
        nx = m.header.nx
        ny = m.header.ny
        nz = m.header.nz
        nsymbt = m.header.nsymbt

    # compute data offset
    offset = 1024 + nsymbt

    # read raw bytes manually
    data = np.fromfile(
        fn,
        dtype=np.float32,   # <-- force desired dtype
        offset=offset
    )

    # reshape manually
    return data.reshape((nz, ny, nx))
```

```python
def read_MRC(fn):
    with mrcfile.open(fn, permissive=True) as data:
        return data.data
```

```python
def array_PCC(original, denoised):
    PCC = scipy.stats.pearsonr(original.flatten(), denoised.flatten())[0]
    print(f"PCC={PCC:.2f}")
```
