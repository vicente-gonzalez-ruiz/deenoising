```python
X = mrcfile.open("../../training_05nm.mrc").data
```

```python
X_REO = mrcfile.open("denoised_vol/training_05nm.mrc").data
```

```python
slice_idx = X.shape[0] // 2

fig, axes = plt.subplots(1, 2, figsize=(10, 10))

# Plot a original slice
_ = axes[0].imshow(X[slice_idx][sliice], cmap='gray', origin='lower')
axes[0].set_title(f'Original Z={slice_idx}')
axes[0].grid(False)

# EO
_ = axes[1].imshow(X_REO[slice_idx][sliice], cmap='gray', origin='lower')
axes[1].set_title(f'Denoised REO Z={slice_idx}')
axes[1].grid(False)

plt.tight_layout()
plt.show()
```
