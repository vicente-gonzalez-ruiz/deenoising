```python
def read_FRC(f):
    x = []
    y = []

    with open(f, 'r') as file:
        for line in file:
            parts = line.split()
            if len(parts) == 2:
                x.append(float(parts[0]))
                y.append(float(parts[1]))
    return x, y
```

```python
x_nearest, y_nearest = read_FRC("/tmp/1")
x_linear, y_linear = read_FRC("/tmp/2")
```

```python
plt.title("EPFL (40 nm)")
plt.xlabel("Normalized Spatial Frequency (cycles/pixel)")
plt.ylabel("Average Self Fourier Ring Correlation")
plt.plot(x_nearest, y_nearest, label="Nearest", color="red")
plt.plot(x_linear, y_linear, label="Linear", color="green")
plt.legend(loc='lower left')
```
