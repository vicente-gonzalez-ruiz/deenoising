```python
import os
import glob
import numpy as np
import mrcfile
import pandas as pd
from collections import namedtuple
#import information_theory as IT
import skimage.metrics
```

```python
def SSIM(original_fn, denoised_fn):
    print(f"Original: {original_fn}")
    print(f"Denoised: {denoised_fn}")
    with mrcfile.open(original_fn, permissive=True) as original, mrcfile.open(denoised_fn, permissive=True) as denoised:
        # Load data as float32 to prevent overflow issues during subtraction
        original_data = original.data.astype(np.float32)
        denoised_data = denoised.data.astype(np.float32)
    SSIM = skimage.metrics.structural_similarity(original_data, denoised_data, data_range=original_data.max() - original_data.min())
    print(f"SSIM={SSIM:.2f}")
```
